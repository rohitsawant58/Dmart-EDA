# Task 3: Descriptive Statistics

```python
import Omkar_statistics
```

```python
Omkar_statistics.cal_stats('dmart Discount Price',df ['Discount Price'])
```

```python
break_even=set(df[df['Discount Price']==0]['MRP'])
```

```python
loss_making=set(df[df['Discount Price']<0]['MRP'])
```

```python
profit_making=set(df[df['Discount Price']>0]['MRP'])
```

```python
rating_billnum=[]
for i in profit_making:
    if i in loss_making or i in break_even:
        pass
    else:
        rating_billnum.append(i)
```

```python
df[df['MRP'].isin(rating_billnum)]['Discount Price'].mean()
```

```python
sns.scatterplot(data=df,x='MRP',y='Discount Price')
plt.title('Rating V/S Discount Price')
plt.show()
```

```python
sns.pairplot(df[['Product Name','MRP','Discount Price','Rating']])
```

```python
is_discountprice=[]
for i in df['Discount Price']:
    if i>0:
        is_discountprice.append('Yes')
    else:
        is_discountprice.append('NO')
```

```python
df['Discount Price']=is_discountprice
```

```python
sns.pairplot(data=df,hue='Discount Price')
```

```python
df['Order Date']=pd.to_datetime(df['Order Date'], format='%d-%m-%Y')
```

```python
df['Delivery Date']=pd.to_datetime(df['Delivery Date'])
```

```python
plt.plot(df.resample('M',on='Order Date')['Rating'].sum())
plt.title('Monthly Rating')
plt.xlabel('Month Year')
plt.ylabel('Total Rating')
plt.show()
```

```python
state_counts=df['State'].value_counts()
```

```python
plt.pie(x=state_counts.values, labels=state_counts.index,autopct='%0.1f%%')
plt.title('State Pie Chart',color='white')
plt.show()
```

# Task 5: Grouped Analysis

```python
category_discountprice=df.groupby('Category')['Discount Price'].mean()
plt.bar(category_discountprice.index,category_discountprice.values)
plt.title('Category Wise Discount Price')
plt.show()
```
